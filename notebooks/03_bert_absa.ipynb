# 03 — BERT ABSA Pipeline

Interactively explore the two-stage BERT ABSA pipeline (aspect extraction + sentiment classification).
Models must already be trained:
```
python -m src.training.train_aspect
python -m src.training.train_sentiment
```

````python
import os, sys
sys.path.append(os.path.abspath('..'))
from src.inference.predict import predict_absa, overall_sentiment, ModelsNotTrainedError
````

````python
examples = [
    "The food was amazing but the service was extremely slow.",
    "The camera quality is excellent, but the battery life is disappointing.",
    "The food wasn't bad.",
    "The food was good but expensive.",
]

for text in examples:
    print(f"Input: {text}")
    try:
        result = predict_absa(text)
        for item in result:
            print(f"  - {item['aspect']:20s} -> {item['sentiment']:10s} (confidence={item['confidence']:.2f})")
        print(f"  Overall: {overall_sentiment(result)}")
    except ModelsNotTrainedError as exc:
        print(f"  {exc}")
    print()
````

## Try your own review

````python
my_review = "The staff were friendly but the room was dirty and small."
try:
    result = predict_absa(my_review)
    for item in result:
        print(item)
    print("Overall:", overall_sentiment(result))
except ModelsNotTrainedError as exc:
    print(exc)
````
