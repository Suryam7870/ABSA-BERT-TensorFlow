# 01 — Data Exploration

Explore the processed ABSA dataset (run `python -m src.data.prepare_data` first).

```python
import os, sys, json
sys.path.append(os.path.abspath('..'))
import pandas as pd
from config import cfg

pd.set_option('display.max_colwidth', 100)
```

## Load processed sentiment data

```python
sentiment_train_path = cfg.SENTIMENT_TRAIN_FILE
if os.path.exists(sentiment_train_path):
    df = pd.read_csv(sentiment_train_path)
    print(f"Loaded {len(df)} rows from {sentiment_train_path}")
    display(df.head(10))
else:
    print(f"File not found: {sentiment_train_path}\nRun `python -m src.data.prepare_data` first.")
    df = pd.DataFrame(columns=['text', 'aspect', 'sentiment'])
```

## Sentiment class distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns

if not df.empty:
    fig, ax = plt.subplots(figsize=(6, 4))
    sns.countplot(x='sentiment', data=df, order=cfg.SENTIMENT_LABELS, ax=ax)
    ax.set_title('Sentiment class distribution (train split)')
    plt.show()
else:
    print("No data loaded yet.")
```

## Aspect term frequency

```python
if not df.empty:
    top_aspects = df['aspect'].str.lower().value_counts().head(20)
    display(top_aspects)
else:
    print("No data loaded yet.")
```

## Review length distribution

```python
if not df.empty:
    df['num_words'] = df['text'].str.split().apply(len)
    fig, ax = plt.subplots(figsize=(6, 4))
    sns.histplot(df['num_words'], bins=20, ax=ax)
    ax.set_title('Review length (words)')
    plt.show()
    print(df['num_words'].describe())
else:
    print("No data loaded yet.")
```

## Load processed aspect-extraction (BIO) data

```python
aspect_train_path = cfg.ASPECT_TRAIN_FILE
if os.path.exists(aspect_train_path):
    aspect_df = pd.read_csv(aspect_train_path)
    print(f"Loaded {len(aspect_df)} rows from {aspect_train_path}")
    example = aspect_df.iloc[0]
    tokens = json.loads(example['tokens'])
    labels = json.loads(example['bio_labels'])
    for tok, lab in zip(tokens, labels):
        print(f"{tok:20s} {lab}")
else:
    print(f"File not found: {aspect_train_path}\nRun `python -m src.data.prepare_data` first.")
```
